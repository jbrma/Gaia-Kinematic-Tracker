```python
import pandas as pd
import numpy as np

df = pd.read_csv('pulsars/ATNF_PulsarCatalogue.csv')

print(f"Total: {len(df)}")
```

```python
df['AGE'] = pd.to_numeric(df['AGE'].replace('*', np.nan), errors='coerce')

age_pulsars = df.dropna(subset=['AGE'])
print(f"Pulsars with age: {len(age_pulsars)}")
```

```python
df_young = age_pulsars.sort_values(by='DIST', ascending=True)

close_pulsars = df_young[df_young['AGE'] < 100000]

columns = ['NAME', 'AGE', 'DIST', 'ASSOC', 'PMTOT', 'VTRANS']

close_pulsars[columns]
```

```python
df['DIST'] = pd.to_numeric(df['DIST'].replace('*', np.nan), errors='coerce')

close_pulsars = df[df['DIST'] < 1.0]
close_pulsars[['NAME', 'DIST', 'AGE']].sort_values(by='DIST')
```
